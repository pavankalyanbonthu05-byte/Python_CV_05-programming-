```python
def minrotate_array(nums,t):
    left = nums
    right = nums
    k = 1
    while k < len(nums):
        left = left[-k:] + left[:-k]
        right = right[k:] + right[:k]
        if left == t or right == t:
            return k
        else:
            k += 1
    return 0
nums=[1,2,3,4,5]
t=[2,3,4,5,1]
print(minrotate_array(nums,t))
```

```python
def minrotate_array(nums,t):
    if nums == t:
        return 0
    n = len(nums)
    c = nums + nums
    for k in range(1,n+1):
        if c[n-k : 2*n-k] == t:
            return k
        if c[k:k+n] == t:
            return k
    return -1
nums=[1,2,3,4,5]
t=[2,3,4,5,1]
print(minrotate_array(nums,t))
```

```python
def count_arrangements(N, R, end):
    dp = [0] * (R + 1)
    dp[1] = 1   # start must be 1

    for _ in range(2, N + 1):
        new_dp = [0] * (R + 1)
        total = sum(dp)

        for j in range(1, R + 1):
            new_dp[j] = total - dp[j]

        dp = new_dp

    return dp[end]
N = 4
R = 4
end = 3
print(count_arrangements(N, R, end))
```

```python
def ballon(w,m):
    ind_w=[]
    r=[]
    c=0
    for i in range(len(w)):
        ind_w.append((w[i],i))
    ind_w.sort()
    for val,i in ind_w:
        if val+c <= m:
            c += val
            r.append(i)
        else:
            break
    return r
w=[20,30,10,50,40]
m=100
print(ballon(w,m))
```
