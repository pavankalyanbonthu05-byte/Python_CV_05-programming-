```python
def sorted_arr(nums):
    evens = sorted([x for x in nums if x % 2 == 0])
    odds = sorted([x for x in nums if x % 2 != 0], reverse=True)

    # Use iterators to pull from sorted lists as we reconstruct
    e_idx, o_idx = 0, 0
    result = []

    for x in nums:
        if x % 2 == 0:
            result.append(evens[e_idx])
            e_idx += 1
        else:
            result.append(odds[o_idx])
            o_idx += 1

    return result

nums = [5, 8, 2, 1, 9]
print(sorted_arr(nums))
```

```python
def binary(n):
    bits=n.bit_length()
    t = n ^ ((1 << bits)-1)
    return t
n = 10
print(binary(n))
```

```python
def to_binary(n):
    b=''
    if n == 0:
        return 0
    while n > 0:
        b = str(n%2)+b
        n = n // 2
    return b
n = 10
print(to_binary(n))
```

```python
def product_div(nums):
    p = 1
    for i in range(len(nums)):
        p *= nums[i]
    r=[]
    for i in range(len(nums)):
        r.append(p // nums[i])
    return r
nums = [1,2,3,4]#works for only non zero elements
print(product_div(nums))
```

```python
def product(nums):
    n = len(nums)
    res = [1] * n
    p=1
    for i in range(len(nums)):
        res[i] = p
        p *= nums[i]
    s =1
    for i in range(len(nums)-1,-1,-1):
        res[i] *= s
        s *= nums[i]
    return res
nums=[1,0,3,4]
print(product(nums))
```

```python
def moving_zeros(nums):
    for i in range(len(nums)):
        for j in range(i,len(nums)):
            if nums[i] == 0 and nums[j] != 0:
                nums[i],nums[j] = nums[j],nums[i]
        if nums[i] != 0:
            i += 1
    return nums
nums=[4,0,5,0,6,0]
print(moving_zeros(nums))
```

```python
def moving_zeros(nums):
    pos = 0
    for i in range(len(nums)):
        if nums[i] != 0:
            nums[pos], nums[i] = nums[i], nums[pos]
            pos += 1
    return nums
nums=[4,0,5,0,6,0]
print(moving_zeros(nums))
```

```python
def max_product(nums):
    if not nums:
        return 0
    res=max(nums)
    curr_max, curr_min = 1,1
    for n in nums:
        if n == 0:
            curr_max, curr_min = 1,1
            continue
        temp = curr_max * n
        curr_max = max(curr_max * n, curr_min * n,n)
        curr_min = min(temp,n*curr_min,n)
        res=max(res,curr_max)
    return res
nums=[2, 3, -2, 4]
print(max_product(nums))
```

```python
def stock_buy_2(nums):
    total_profit = 0
    i = 0
    for j in range(i+1,len(nums)):
        if nums[j] - nums[i] > 0:
            total_profit += nums[j] - nums[i]
        i += 1
    return total_profit
nums=[1,5,3,8,12]
print(stock_buy_2(nums))
```

```python
def equilibirium(nums):
    total_sum = sum(nums)
    left_sum = 0
    for i,val in enumerate(nums):
        right_sum = total_sum - left_sum - val
        if left_sum == right_sum:
            return val,i
        left_sum += val
    return -1
nums=[1, 7, 3, 6, 5, 6]
print(equilibirium(nums))
```
