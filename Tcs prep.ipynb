```python
'''Eqilibirium of array'''
def equilibirium(nums):
    left = 0
    total = sum(nums)
    for i in range(len(nums)):
        right = total - left - nums[i]
        if left == right:
            return i
        left += nums[i]
    return 'none'
nums=[1, 2, 3, 4, -10, 10]
print(equilibirium(nums))
```

```python
def maxArea(self, height: List[int]) -> int:
        left = 0
        right = len(height)-1
        max_w = 0
        while left < right:
            w = right - left
            h = min(height[left],height[right])
            area = w*h
            max_w = max(max_w,area)
            if height[left] < height[right]:
                left += 1
            else:
                right -= 1
        return max_w
```

```python
def findMedianSortedArrays(self, nums1: List[int], nums2: List[int]) -> float:
        m = nums1 + nums2
        m.sort()
        n = len(m)
        if n % 2 != 0:
            return (m[n // 2])
        else:
            mid1 = m[n // 2-1]
            mid2 = m[n // 2]
            return (mid1 + mid2) / 2
```

```python
def trapping_rain(n):
    l=0
    r=len(n)-1
    max_l = max_r = 0
    total_w = 0
    while l < r:
        if n[l] < n[r]:
            if n[l] >= max_l:
                max_l = n[l]
            else:
                total_w += max_l - n[l]
            l += 1
        else:
            if n[r] >= max_r:
                max_r = n[r]
            else:
                total_w += max_r - n[r]
            r -= 1
    return total_w
n=list(map(int,input().split()))
print(trapping_rain(n))
```

```python
def secound_largest(nums):
    first=float('-inf')
    sec=float('-inf')
    for i in range(len(nums)):
        if nums[i] > first:
            sec = first
            first = nums[i]
        elif nums[i] > sec and nums[i] != first:
            sec = nums[i]
    return sec
nums=list(map(int,input().split()))
print(secound_largest(nums))
```
